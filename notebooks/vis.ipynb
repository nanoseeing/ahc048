```python
from dataclasses import dataclass


def data_input():
    N, K, H, T, D = map(int, input().split())
    own_colors = []
    for _ in range(K):
        c, m ,y = map(float, input().split())
        own_colors.append((c, m, y))
    
    target_colors = []
    for _ in range(H):
        c, m, y = map(float, input().split())
        target_colors.append((c, m, y))
    
    return N, K, H, T, D, own_colors, target_colors

def data_input_from_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
    
    N, K, H, T, D = map(int, lines[0].strip().split())

    own_colors = []
    for i in range(1, K + 1):
        c, m, y = map(float, lines[i].strip().split())
        own_colors.append((c, m, y))

    target_colors = []
    for i in range(K + 1, K + H + 1):
        c, m, y = map(float, lines[i].strip().split())
        target_colors.append((c, m, y))
    return N, K, H, T, D, own_colors, target_colors
```

```python
from pathlib import Path

file_dir = Path("./tools/in")
file_num = 100

t_list = []
d_list = []
for i in range(file_num):
    file_path = Path(file_dir) / f"{i:04d}.txt"
    n, k, h, t, d, own_colors, target_colors = data_input_from_file(file_path)
    t_list.append(t)
    d_list.append(d)
    
```

```python
import matplotlib.pyplot as plt

# histogram of t values
plt.hist(d_list, bins=20, alpha=0.7, color='blue', edgecolor='black')
plt.title('Histogram of T values')
plt.xlabel('T value')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
