```python
import matplotlib.pyplot as plt
```

```python
MAX_RATE = 19
```

```python
vals = []
for i in range(1, MAX_RATE+1):
    for j in range(1, i):
        # vals.append((j/i, (j, i)))
        vals.append(j/i)

vals.append(0.0)
vals.append(1.0)
```

```python
vals2 = []
for i in range(2, MAX_RATE*2+1):
    for j in range(1, i):
        for i2 in range(max(2, j), j + (MAX_RATE*2 - i) + 1):
            for j2 in range(1, i2):
                vals2.append(j/i * j2/i2 * 1.25)
                # print(f"j/i: {j}/{i}, j2/i2: {j2}/{i2}, value: {j/i * j2/i2}")
vals2.append(0.0)
vals2.append(1.25)

vals4 = []
for i in range(2, MAX_RATE*2+1):
    for j in range(1, i):
        for i2 in range(max(2, j), j + (MAX_RATE*2 - i) + 1):
            for j2 in range(1, i2):
                vals4.append(j/i * j2/i2 * 0.25)
                # print(f"j/i: {j}/{i}, j2/i2: {j2}/{i2}, value: {j/i * j2/i2}")
vals4.append(0.0)
vals4.append(0.25)
```

```python
vals3 = []

tmp_vals1 = []
tmp_vals2 = []
for i in range(1, MAX_RATE+1):
    for j in range(1, i):
        # vals.append((j/i, (j, i)))
        tmp_vals1.append(j/i * 0.3)
        tmp_vals2.append(j/i * 1.52)

for v1 in tmp_vals1:
    for v2 in tmp_vals2:
        vals3.append(v1 + v2)
```

```python
vals5 = []
for i in range(1, MAX_RATE+1):
    for j in range(1, i):
        # vals.append((j/i, (j, i)))
        vals5.append(j/i * 0.32)

for i in range(1, MAX_RATE+1):
    for j in range(1, i):
        # vals.append((j/i, (j, i)))
        vals5.append(j/i * 1.28)
```

```python
vals6 = []
for i in range(1, MAX_RATE*2+1):
    for j in range(1, i):
        # vals.append((j/i, (j, i)))
        vals6.append(j/i)

vals6.append(0.0)
vals6.append(1.0)
```

```python
vals = sorted(set(vals))
vals2 = sorted(set(vals2))
vals3 = sorted(set(vals3))
vals4 = sorted(set(vals4))
vals5 =sorted(set(vals5))
vals6 = sorted(set(vals6))
print(len(vals), len(vals2), len(vals3), len(vals4), len(vals5), len(vals6))
```

```python
fig = plt.figure(figsize=(20, 3))
plt.scatter(vals, [0] * len(vals),s=1, label='Values', color='red')
plt.scatter(vals2, [0.1] * len(vals2), s=1, label='Values', color='green')
plt.scatter(vals3, [0.2] * len(vals3), s=1, label='Values', color='blue')
plt.scatter(vals4, [0.3] * len(vals4), s=1, label='Values', color='green')
plt.scatter(vals5, [0.5] * len(vals5), s=1, label='Values', color='black')
plt.scatter(vals6, [0.6] * len(vals6), s=1, label='Values', color='orange')
plt.show()
```
