```python
import subprocess

file_num = 100

results = []
weights = []
for i in range(file_num):
    print(f"=============== Processing file {i:04d} ===============")
    file_path = f"../tools/in/{i:04d}.txt"
    try:
        with open(file_path, "r") as f:
            process = subprocess.Popen(
                ["../a.out"],
                stdin=f,
                stdout=subprocess.PIPE,
                stderr=subprocess.PIPE,
                text=True
            )

            stdout, stderr = process.communicate()
            lines = stdout.splitlines()
            
            result = map(float,lines[0].strip().split())
            results.append(list(result))
            
            # weight = map(float, lines[1].strip().split())
            # weights.append(list(weight))

    except Exception as e:
        print(f"Error processing {file_path}: {e}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5, 20))
plt.boxplot(results, vert=False)
plt.xlabel("Values")
plt.title("Boxplot of Results from a.out")
plt.show()
```

```python
plt.hist(results[11], bins=50)
plt.show()
```

```python
avg_resutls = [sum(res) / len(res) for res in results]
inds = list(range(len(avg_resutls)))

plt.figure(figsize=(20, 5))
plt.bar(inds, avg_resutls)
plt.show()
```
